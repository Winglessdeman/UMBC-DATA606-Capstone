# NHANES Diabetes Risk Data Exploration

In this notebook, I explore NHANES data from the following survey cycles:
- 2011–2012 
- 2013–2014 
- 2015–2016 
- 2017–2018 
It contains 32 NHANES .xpt source files across the survey cycle with the combined raw file size of ~ 65.20 MB  
Each component contains a different type of participant information

`DEMO`: Demographics - Age, sex, race/ethnicity, education, income, and survey design variables 

`GHB` : Glycohemoglobin - Laboratory-measured HbA1c; source of the machine-learning target 

`BMX` : Body Measures - BMI, weight, height, waist circumference, and related measurements 

`DIQ` : Diabetes Questionnaire - Self-reported diabetes diagnosis and related diabetes information 

`PAQ` : Physical Activity - Physical activity, exercise, and sedentary behavior 

`SMQ` : Smoking - Cigarette Use - Lifetime and current cigarette-smoking behavior 

`SLQ` : Sleep Disorders - Sleep duration and selected sleep-related information 

`BPQ` : Blood Pressure & Cholesterol Questionnaire - Self-reported hypertension and cholesterol-related information 

These files are in SAS transport format where each survey cycle uses suffix in the filename to represent the year.

G: 2011-2012, H: 2013–2014, I: 2015–2016, J: 2017–2018

# 2. Importing the Data

```python
# Setting up the path 
data_dir = Path("../data")

#  NHANES components to be used
components = [
    "DEMO",  # Demographics
    "GHB",   # Glycohemoglobin / HbA1c
    "BMX",   # Body Measures
    "DIQ",   # Diabetes Questionnaire
    "PAQ",   # Physical Activity
    "SMQ",   # Smoking - Cigarette Use
    "SLQ",   # Sleep Disorders
    "BPQ"    # Blood Pressure & Cholesterol Questionnaire
]

# NHANES survey cycles
cycles = {
    "G": {
        "year": "2011",
        "cycle": "2011-2012"
    },
    "H": {
        "year": "2013",
        "cycle": "2013-2014"
    },
    "I": {
        "year": "2015",
        "cycle": "2015-2016"
    },
    "J": {
        "year": "2017",
        "cycle": "2017-2018"
    }
}

#  Downloading all required NHANES files
for suffix, cycle_info in cycles.items():

    year = cycle_info["year"]
    cycle_name = cycle_info["cycle"]

    print("\n" + "=" * 60)
    print(f"NHANES {cycle_name}")
    print("=" * 60)

    # going through each data component.
    for component in components:

        # Building the expected filename i.e DEMO_G.xpt
        filename = f"{component}_{suffix}.xpt"

        # local path where the file should be saved.
        output_path = data_dir / filename

        # Check if the file already exists
   

        if output_path.exists():

            # file size for reference.
            size_mb = output_path.stat().st_size / (1024 ** 2)

            print(
                f"EXISTS      {filename:<12} "
                f"{size_mb:.3f} MB"
            )

            # Skipping  if the file is already present.
            continue


        # official CDC download URL
        url = (
            f"https://wwwn.cdc.gov/Nchs/Data/Nhanes/"
            f"Public/{year}/DataFiles/{filename}"
        )

        #print(f"Downloading {filename} ...")

        try:

            # Requesting the files from the CDC website.
            response = requests.get(
                url,
                timeout=60
            )

            # Raising an error if the server returned
            response.raise_for_status()


            # checking CDC returns an XPT file and not an HTML error page.
            content = response.content

            first_bytes = content.lstrip().lower()[:100]

            if (
                first_bytes.startswith(b"<!doctype")
                or first_bytes.startswith(b"<html")
            ):
                print(
                    f"ERROR       {filename:<12} "
                    f"CDC returned a webpage instead of XPT data."
                )
                continue

            # Saving the downloaded file
            output_path.write_bytes(content)

            # Calculating file size 
            size_mb = output_path.stat().st_size / (1024 ** 2)

            print(
                f"DOWNLOADED  {filename:<12} "
                f"{size_mb:.3f} MB"
            )

        except requests.RequestException as error:

            print(
                f"FAILED      {filename:<12} "
                f"{error}"
            )

# Create list of all expected NHANES filenames
expected_files = [
    f"{component}_{suffix}.xpt"
    for suffix in cycles
    for component in components
]
        
#total file size           
active_files = [
    data_dir / filename
    for filename in expected_files
    if (data_dir / filename).exists()
]

total_size_mb = sum(
    file.stat().st_size
    for file in active_files
) / (1024 ** 2)

print(
    f"\nTotal size of active NHANES files: "
    f"{total_size_mb:.3f} MB"
)
```

```python
# Special NHANES response codes for the selected variables
# These codes represent responses such as Refused or Don't know.

special_code_map = {

    # Demographics
    "education": [7, 9],

    # Diabetes questionnaire
    "diabetes_report": [7, 9],

    # Physical activity
    "vigorous_activity": [7, 9],
    "vigorous_days": [77, 99],
    "vigorous_minutes": [7777, 9999],

    "moderate_activity": [7, 9],
    "moderate_days": [77, 99],
    "moderate_minutes": [7777, 9999],

    "sedentary_minutes": [7777, 9999],

    # Smoking
    "smoked_100_cigarettes": [7, 9],
    "current_smoking": [7, 9],

    # Sleep
    "sleep_hours": [77, 99],

    # Blood pressure / cholesterol
    "hypertension": [7, 9],
    "high_cholesterol": [7, 9]
}

```

```python
def clean_special_codes(df):
    """
    Replace NHANES special response codes with NaN
    for the selected analysis variables.
    """

    df = df.copy()

    # Replace variable-specific special codes
    for column, codes in special_code_map.items():

        if column in df.columns:
            df[column] = df[column].replace(codes, np.nan)

    # PAD680 can contain a tiny floating-point number
    # after importing the SAS XPT file.
    # Values extremely close to zero are treated as 0.
    if "sedentary_minutes" in df.columns:

        df.loc[
            df["sedentary_minutes"].abs() < 1e-10,
            "sedentary_minutes"
        ] = 0

    return df
```

```python
def clean_four_cycles(g, h, i, j):
    """
    Apply special-code cleaning to the same NHANES
    component across all four survey cycles.
    """

    return (
        clean_special_codes(g),
        clean_special_codes(h),
        clean_special_codes(i),
        clean_special_codes(j)
    )
```

###  Creating Physical-Activity Features

The NHANES Physical Activity Questionnaire records recreational activity using
multiple related variables.

For vigorous recreational activity:

- vigorous_activity indicates whether the participant performs vigorous
  recreational activity.
- vigorous_days records the number of days per week.
- vigorous_minutes records the number of minutes performed on a typical day.

For moderate recreational activity:

- moderate_activity indicates whether the participant performs moderate
  recreational activity.
- moderate_days records the number of days per week.
- moderate_minutes records the number of minutes performed on a typical day.

To make these variables easier to use in exploratory analysis and machine learning, weekly activity measures will be calculated as:

`days per week × minutes per day`

Participants who report that they do not perform the activity will receive
0 minutes per week.

If a participant reports performing the activity but the number of days or minutes is missing, the derived weekly value will remain missing rather than being assumed to be zero.

sedentary_minutes will remain as minutes per day.

```python
def create_activity_features(df):
    """
    Create weekly vigorous and moderate recreational
    physical-activity features.
    """

    df = df.copy()

    # Vigorous recreational activity

    #  missing values
    df["vigorous_minutes_week"] = np.nan

    # PAQ650 = 2 means participant reported NO vigorous activity
    df.loc[
        df["vigorous_activity"] == 2,
        "vigorous_minutes_week"
    ] = 0

    # PAQ650 = 1 means YES.
    # Calculating weekly minutes only when both days and minutes are available.
    vigorous_yes = (
        (df["vigorous_activity"] == 1)
        & df["vigorous_days"].notna()
        & df["vigorous_minutes"].notna()
    )

    df.loc[
        vigorous_yes,
        "vigorous_minutes_week"
    ] = (
        df.loc[vigorous_yes, "vigorous_days"]
        * df.loc[vigorous_yes, "vigorous_minutes"]
    )

    # Moderate recreational activity

    df["moderate_minutes_week"] = np.nan

    # PAQ665 = 2 means participant reported NO moderate activity
    df.loc[
        df["moderate_activity"] == 2,
        "moderate_minutes_week"
    ] = 0

    # Calculating weekly minutes for participants reporting YES
    moderate_yes = (
        (df["moderate_activity"] == 1)
        & df["moderate_days"].notna()
        & df["moderate_minutes"].notna()
    )

    df.loc[
        moderate_yes,
        "moderate_minutes_week"
    ] = (
        df.loc[moderate_yes, "moderate_days"]
        * df.loc[moderate_yes, "moderate_minutes"]
    )

    return df
```

## 6. Combining Survey Cycles and Merging Components


After selecting, cleaning, and standardizing the variables, the four NHANES survey cycles can be combined.

First, the same component from each survey cycle will be stacked vertically. For example, the four demographic datasets will be combined into one demographic dataset.

After each component has been combined across cycles, the components will be merged using `SEQN`, the unique NHANES participant identifier.

The demographic dataset will serve as the base dataset so that participants are retained even when information from another component is missing.

```python
# Combine the four survey cycles within each component

demo_all = pd.concat(
    [
        demo_g_selected,
        demo_h_selected,
        demo_i_selected,
        demo_j_selected
    ],
    ignore_index=True
)

ghb_all = pd.concat(
    [
        ghb_g_selected,
        ghb_h_selected,
        ghb_i_selected,
        ghb_j_selected
    ],
    ignore_index=True
)

bmx_all = pd.concat(
    [
        bmx_g_selected,
        bmx_h_selected,
        bmx_i_selected,
        bmx_j_selected
    ],
    ignore_index=True
)

diq_all = pd.concat(
    [
        diq_g_selected,
        diq_h_selected,
        diq_i_selected,
        diq_j_selected
    ],
    ignore_index=True
)

paq_all = pd.concat(
    [
        paq_g_selected,
        paq_h_selected,
        paq_i_selected,
        paq_j_selected
    ],
    ignore_index=True
)

smq_all = pd.concat(
    [
        smq_g_selected,
        smq_h_selected,
        smq_i_selected,
        smq_j_selected
    ],
    ignore_index=True
)

slq_all = pd.concat(
    [
        slq_g_selected,
        slq_h_selected,
        slq_i_selected,
        slq_j_selected
    ],
    ignore_index=True
)

bpq_all = pd.concat(
    [
        bpq_g_selected,
        bpq_h_selected,
        bpq_i_selected,
        bpq_j_selected
    ],
    ignore_index=True
)

print("DEMO:", demo_all.shape)
print("GHB :", ghb_all.shape)
print("BMX :", bmx_all.shape)
print("DIQ :", diq_all.shape)
print("PAQ :", paq_all.shape)
print("SMQ :", smq_all.shape)
print("SLQ :", slq_all.shape)
print("BPQ :", bpq_all.shape)
```

```python
combined_components = {
    "DEMO": demo_all,
    "GHB": ghb_all,
    "BMX": bmx_all,
    "DIQ": diq_all,
    "PAQ": paq_all,
    "SMQ": smq_all,
    "SLQ": slq_all,
    "BPQ": bpq_all
}

for name, df in combined_components.items():

    duplicates = df["SEQN"].duplicated().sum()

    print(
        f"{name:<5} rows: {len(df):>6} "
        f"duplicate SEQN: {duplicates}"
    )
```

```python
#  demographics as the base participant dataset

nhanes = demo_all.copy()

# Merge laboratory HbA1c
nhanes = nhanes.merge(
    ghb_all,
    on="SEQN",
    how="left"
)

# Merge body measurements
nhanes = nhanes.merge(
    bmx_all,
    on="SEQN",
    how="left"
)

# Merge diabetes questionnaire
nhanes = nhanes.merge(
    diq_all,
    on="SEQN",
    how="left"
)

# Merge physical activity
nhanes = nhanes.merge(
    paq_all,
    on="SEQN",
    how="left"
)

# Merge smoking
nhanes = nhanes.merge(
    smq_all,
    on="SEQN",
    how="left"
)

# Merge sleep
nhanes = nhanes.merge(
    slq_all,
    on="SEQN",
    how="left"
)

# Merge blood pressure and cholesterol
nhanes = nhanes.merge(
    bpq_all,
    on="SEQN",
    how="left"
)

print("Final merged shape:", nhanes.shape)
```
